```python
import time

def timer(func):
    
    def wraper():
        start = time.time()
        func()
        end = time.time() - start
        return f"time spend: {end}"
    
    return wraper

@timer
def cap():
    total = 0
    for i in range(1, 80000000):
        total += 1
    return total


cap()  
```
